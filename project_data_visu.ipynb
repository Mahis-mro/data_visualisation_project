## 2.Data Cleaning


### 2.1 Feature Selection

The original dataset contains 53 columns, but not all of them are relevant to our analysis.

To simplify the dataset and improve processing efficiency, we selected 12 columns directly related to our research questions:

- **Inspection identification:** `Numero_inspection` and `SIRET` identify inspections and establishments.
- **Establishment information:** Names and activity types allow comparisons between restaurants, bakeries, and other food businesses.
- **Sanitary evaluation:** `Synthese_eval_sanit` provides the inspection results.
- **Inspection context and date:** These variables allow us to analyze differences between inspection types and examine temporal patterns.
- **Geographic information:** Region, department, city, postal code, and coordinates support geographical comparisons and interactive maps.

The remaining 41 columns were excluded because they are not necessary for our dashboard.

This operation reduces the dataset from 53 to 12 columns while preserving all 71,326 observations. The original dataset remains unchanged.


```python
df_clean = df.copy()
columns_to_keep = [
    "Numero_inspection",
    "SIRET",
    "APP_Libelle_etablissement",
    "APP_Libelle_activite_etablissement",
    "Synthese_eval_sanit",
    "Date_inspection",
    "Contexte_inspection",
    "reg_name",
    "dep_name",
    "com_name",
    "Code_postal",
    "geores"
]

df_clean = df_clean[columns_to_keep].copy()
```

```python
# new dimension

df_clean.shape
```


### 2.2 Handling Missing Values

After selecting the relevant columns, we identified several missing values.

- **Sanitary evaluation:** One inspection has no sanitary rating. This observation is excluded because the evaluation is essential to our analysis.
- **Establishment activity:** The missing activity value is replaced with "Unknown" to preserve the observation.
- **Geographic information:** Missing region, department and city names are labeled "Unknown". These inspections remain useful for national statistics but will be excluded from geographic comparisons when necessary.
- **Geographic coordinates:** Missing coordinates are preserved as null values. They will only be excluded from map visualizations.

This approach avoids removing valid inspections unnecessarily and preserves as much information as possible.


```python
# Remove inspections without a sanitary evaluation
df_clean = df_clean.dropna(subset=["Synthese_eval_sanit"]).copy()
```

```python
# Replace missing categorical information
columns_unknown = [
    "APP_Libelle_activite_etablissement",
    "reg_name",
    "dep_name",
    "com_name"
]

df_clean[columns_unknown] = df_clean[columns_unknown].fillna("Unknown")
```

```python
# Verify missing values after cleaning
df_clean.isnull().sum().sort_values(ascending=False)
```


### 2.3 Data Type Conversion

To prepare the dataset for analysis and visualization, we converted the inspection dates and geographic coordinates into appropriate data types.

- **Inspection dates:** The `Date_inspection` column is converted into datetime format to enable temporal analysis and monthly comparisons.
- **Geographic coordinates:** The `geores` column contains latitude and longitude values stored as text. These coordinates are separated and converted into numeric values to support interactive maps.

Missing geographic coordinates remain as null values, ensuring that no artificial locations are introduced.

These transformations make the dataset suitable for temporal and geographic visualizations in Streamlit.


```python

# Convert inspection dates to datetime format

df_clean["Date_inspection"] = pd.to_datetime(
    df_clean["Date_inspection"].str[:10],
    errors="coerce"
)

df_clean["Date_inspection"].dtype
```

```python
# converting coordinates

coordinates = df_clean["geores"].str.split(",", n=1, expand=True)

df_clean["latitude"] = pd.to_numeric(
    coordinates[0],
    errors="coerce"
)

df_clean["longitude"] = pd.to_numeric(
    coordinates[1],
    errors="coerce"
)

df_clean[["geores", "latitude", "longitude"]].head()
```

```python

# conversion verification

print("Data types:")
print(df_clean[["Date_inspection", "latitude", "longitude"]].dtypes)

print("\nMissing values:")
print(df_clean[["Date_inspection", "latitude", "longitude"]].isnull().sum())

print("\nInvalid coordinates:")
invalid_coords = (
    df_clean["latitude"].notna()
    & df_clean["longitude"].notna()
    & (
        ~df_clean["latitude"].between(-90, 90)
        | ~df_clean["longitude"].between(-180, 180)
    )
)

print(invalid_coords.sum())
```

## 3. Data Analysis (EDA)

The objective of this section is to analyze the results of official food safety inspections in France and identify patterns related to establishment activities, inspection contexts, geographic locations, and time.

The analysis focuses on both the number of inspections and the proportion of unfavorable sanitary evaluations to avoid misleading comparisons between groups of different sizes.

### 3.1 Sanitary Evaluation Analysis

This first analysis examines the overall distribution of sanitary evaluations.

The objective is to understand how inspected food establishments are rated and determine the proportion of inspections requiring improvements or urgent corrective measures.

```python
# Count the number of inspections for each sanitary evaluation

ratings = df_clean["Synthese_eval_sanit"].value_counts()

display(ratings)
```

```python
# Calculate the percentage of inspections for each sanitary evaluation

ratings_percentage = (
    df_clean["Synthese_eval_sanit"]
    .value_counts(normalize=True) * 100
)

display(ratings_percentage.round(2))
```

```python
# Identify unfavorable inspections and calculate their overall percentage

df_clean["unfavorable"] = df_clean["Synthese_eval_sanit"].isin([
    "A améliorer",
    "A corriger de manière urgente"
])

print("Unfavorable inspections:", df_clean["unfavorable"].sum())
print("Unfavorable rate:",
      round(df_clean["unfavorable"].mean() * 100, 2), "%")
```


**Interpretation:**

Among the 71,325 inspections analyzed, 93.68% received a satisfactory or very satisfactory evaluation, while 6.32% (4,510 inspections) required improvements or urgent corrective measures. Although most inspections received favorable evaluations, these results should not be generalized to all food establishments in France, as inspections may be targeted or conducted as follow-ups. This raises an important question: are unfavorable evaluations equally distributed across different types of food establishments?



### 3.2 Establishment Type Analysis

This analysis compares sanitary evaluation results across different types of food establishments. The objective is to identify which activities have the highest proportions of unfavorable inspections and determine whether these differences are related to the number of inspections conducted. To avoid misleading rankings based on very small samples, only activities with at least 100 inspections are included in the unfavorable rate ranking.


```python

# Identify the 15 most frequently inspected establishment activities

top_activities = (
    df_clean["APP_Libelle_activite_etablissement"]
    .value_counts()
    .head(15)
)

display(top_activities)
```

```python

# Count unfavorable inspections for each establishment activity

unfavorable_by_activity = (
    df_clean.groupby("APP_Libelle_activite_etablissement")["unfavorable"]
    .sum()
    .sort_values(ascending=False)
)

display(unfavorable_by_activity.head(15))
```

```python

# Calculate inspection counts and unfavorable rates by establishment activity

activity_analysis = (
    df_clean.groupby("APP_Libelle_activite_etablissement")
    .agg(
        total_inspections=("Numero_inspection", "count"),
        unfavorable_inspections=("unfavorable", "sum"),
        unfavorable_rate=("unfavorable", "mean")
    )
    .reset_index()
)

activity_analysis["unfavorable_rate"] = (
    activity_analysis["unfavorable_rate"] * 100
)

# Display the 15 most inspected establishment activities

display(
    activity_analysis
    .sort_values("total_inspections", ascending=False)
    .head(15)
    .round(2)
)
```

```python

# Rank establishment activities with at least 100 inspections by unfavorable rate

activity_ranking = (
    activity_analysis[
        activity_analysis["total_inspections"] >= 100
    ]
    .sort_values("unfavorable_rate", ascending=False)
)

display(activity_ranking.head(15).round(2))
```

**Interpretation:**

Restaurants account for the largest number of inspections (33,165) and unfavorable evaluations (2,194). However, their unfavorable rate is 6.62%, which is lower than that of bakeries and pastry shops (9.01%) and butcher shops (8.47%). Among activities with at least 100 inspections, cheese-aging facilities have the highest unfavorable rate (11.54%), although this result is based on only 130 inspections. These findings show that the number of unfavorable inspections alone can be misleading. Comparing percentages provides a better understanding of the differences between establishment activities. However, inspection frequency and selection criteria may vary between activities, so these results do not necessarily reflect the overall sanitary conditions of every establishment in each category.



### 3.3 Inspection Context Analysis

This analysis examines whether sanitary evaluation results differ depending on the context of the inspection. The objective is to compare the number of inspections and unfavorable evaluation rates across inspection contexts. This comparison may help explain some differences observed in previous analyses, since inspections are not necessarily conducted under the same circumstances.


```python

# Count inspections for each inspection context

inspection_contexts = df_clean["Contexte_inspection"].value_counts()

display(inspection_contexts)
```

```python
# Compare inspection contexts by unfavorable evaluation rate

context_analysis = (
    df_clean.groupby("Contexte_inspection")
    .agg(
        total_inspections=("Numero_inspection", "count"),
        unfavorable_inspections=("unfavorable", "sum"),
        unfavorable_rate=("unfavorable", "mean")
    )
    .reset_index()
)

context_analysis["unfavorable_rate"] *= 100

display(
    context_analysis
    .sort_values("unfavorable_rate", ascending=False)
    .round(2)
)
```

**Interpretation:**

Unfavorable rates vary depending on the inspection context. Complaints and reports have the highest rate (19.98%), followed by foodborne illness investigations (18.79%) and targeted inspections (18.41%). Scheduled inspections have a lower rate (6.90%), while follow-up inspections reach only 1.54%. These differences may reflect how establishments are selected for inspection rather than the direct effect of the inspection context.


### 3.4 Geographic Analysis

This analysis examines geographic differences in food safety inspection results across French regions. The objective is to compare inspection volumes and unfavorable evaluation rates between regions. Regions with missing geographic information are excluded from this comparison, but their inspections remain included in the national analysis.


```python

# Count the number of inspections in each region

regional_inspections = (
    df_clean[df_clean["reg_name"] != "Unknown"]
    .groupby("reg_name")
    .size()
    .sort_values(ascending=False)
)

display(regional_inspections)
```

```python

# Calculate inspection counts and unfavorable rates for each region

regional_analysis = (
    df_clean[df_clean["reg_name"] != "Unknown"]
    .groupby("reg_name")
    .agg(
        total_inspections=("Numero_inspection", "count"),
        unfavorable_inspections=("unfavorable", "sum"),
        unfavorable_rate=("unfavorable", "mean")
    )
    .reset_index()
)

regional_analysis["unfavorable_rate"] *= 100

display(
    regional_analysis
    .sort_values("unfavorable_rate", ascending=False)
    .round(2)
)
```

```python

# Check how many inspections have missing regional information

missing_regions = (df_clean["reg_name"] == "Unknown").sum()

print("Inspections with unknown region:", missing_regions)
print("Percentage:", round(missing_regions / len(df_clean) * 100, 2), "%")
```


**Interpretation:**

Unfavorable inspection rates vary considerably across French regions. Mayotte (35.71%) and Guadeloupe (25.91%) have the highest rates, while Normandy reaches 9.74% and Île-de-France 6.34%. However, some territories have relatively few inspections, making their rates less reliable for comparison. Additionally, 2.49% of inspections lack regional information and are excluded from this analysis. These differences should not be interpreted as direct rankings of regional food safety, since inspection practices and selection criteria may vary.



### 3.5 Temporal Analysis

This analysis examines how food safety inspections and sanitary evaluation results change over time. The objective is to identify monthly variations in inspection volumes and unfavorable evaluation rates. Since the dataset covers a limited period, the first and last months may be incomplete. These limitations must be considered when interpreting temporal trends.


```python

# Check the inspection date range and the number of inspections per month

print("First inspection:", df_clean["Date_inspection"].min())
print("Last inspection:", df_clean["Date_inspection"].max())

monthly_counts = (
    df_clean.groupby(df_clean["Date_inspection"].dt.to_period("M"))
    .size()
    .sort_index()
)

display(monthly_counts)
```

```python

# Calculate monthly inspection counts and unfavorable evaluation rates

monthly_analysis = (
    df_clean.groupby(df_clean["Date_inspection"].dt.to_period("M"))
    .agg(
        total_inspections=("Numero_inspection", "count"),
        unfavorable_inspections=("unfavorable", "sum"),
        unfavorable_rate=("unfavorable", "mean")
    )
    .reset_index()
)

monthly_analysis["unfavorable_rate"] *= 100

# Display monthly results with two decimal places for the unfavorable rate

display(
    monthly_analysis.style.format({
        "unfavorable_rate": "{:.2f}"
    })
)
```

**Interpretation:**

The unfavorable inspection rate increased from 2.67% in November 2025 to 15.47% in September 2026, while inspection volumes fluctuated throughout the period. October 2025 and October 2026 are only partially covered by the dataset. The 0.51% unfavorable rate in October 2026 should therefore be interpreted with caution. This trend does not necessarily indicate worsening food safety, as inspection selection and reporting delays may influence the results.


### 3.6 Key Findings

The analysis highlights five main findings:

1. **Sanitary evaluations:** 93.68% of inspections received satisfactory or very satisfactory ratings, while 6.32% (4,510 inspections) required improvements or urgent corrective measures.

2. **Establishment activities:** Restaurants account for the largest number of unfavorable inspections (2,194), but bakeries and pastry shops (9.01%) and butcher shops (8.47%) have higher unfavorable rates than restaurants (6.62%).

3. **Inspection contexts:** Complaints and reports have a 19.98% unfavorable rate, compared with 6.90% for scheduled inspections and 1.54% for follow-up inspections.

4. **Geographic differences:** Unfavorable rates vary considerably across regions, with Mayotte reaching 35.71% and Guadeloupe 25.91%. However, these results should be interpreted cautiously because inspection volumes and selection practices differ.

5. **Temporal patterns:** Unfavorable rates reached 15.47% in September 2026, compared with 2.67% in November 2025. This variation may be influenced by inspection selection, incomplete months, and reporting delays.

**Limitations:**

The dataset covers inspections from October 23, 2025, to October 21, 2026, and does not represent every food establishment in France. Results may be influenced by targeted inspections, unequal inspection volumes, geographic coverage, and publication delays.
